```python
def load_image():
    root = tk.Tk()
    root.withdraw()
    file_path = filedialog.askopenfilename(
        title="Válassz egy képet", 
        filetypes=[("Képfájlok", "*.jpg *.png *.jpeg")]
    )
    if not file_path:
        return None
    img = cv2.imread(file_path)
    return cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

ref_img = load_image()
test_img = load_image()
```
